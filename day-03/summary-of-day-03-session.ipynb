## 🛠️ Technical Setup

### Neovim installation

| Command | What it does |
|---|---|
| `nvim` | Launch Neovim |
| `nvim <directory_name>` | Open a directory |
| `nvim` (run inside the directory) | Open the current directory |

### LazyVim starter installation

**1. Back up any existing Neovim files** in `~/.config/nvim` (skip this if the directory doesn't exist):

```bash
mv ~/.config/nvim{,.bak}
```

**2. Clone the [LazyVim starter](https://github.com/LazyVim/starter) into `~/.config/nvim`:**

```bash
git clone https://github.com/LazyVim/starter ~/.config/nvim
```

**3. Remove the `.git` folder** from the cloned configuration:

```bash
rm -rf ~/.config/nvim/.git
```

**4. Launch Neovim** to download and install the bundled plugins automatically:

```bash
nvim
```

### Plugin installation process

1. Create a new `.lua` file in the `lua/plugins/` directory (for example, `lazygit.lua` or `oil.lua`).
2. Copy the plugin configuration from GitHub, and make sure the file **returns a Lua table** with a `return` statement.
3. Restart Neovim so it detects and installs the new plugin.

```lua
-- ~/.config/nvim/lua/plugins/lazygit.lua
return {
  "kdheepak/lazygit.nvim",
  cmd = { "LazyGit" },
  dependencies = { "nvim-lua/plenary.nvim" },
  keys = {
    { "<leader>lg", "<cmd>LazyGit<cr>", desc = "LazyGit" },
  },
}
```

```lua
-- ~/.config/nvim/lua/plugins/oil.lua
return {
  "stevearc/oil.nvim",
  lazy = false,
  opts = {},
  dependencies = { "nvim-tree/nvim-web-devicons" },
  keys = {
    { "-", "<cmd>Oil<cr>", desc = "Open parent directory" },
  },
}
```

> 💡 `<leader>` is `Space` in LazyVim, so `<leader>lg` means `Space` then `l` then `g`.

## ⌨️ Key Bindings & Tools

### File explorer

| Key | Action |
|---|---|
| `Space` `e` | Open or toggle the navigation panel |
| `h` / `j` / `k` / `l` | Navigate |

### LazyGit

| Key | Action |
|---|---|
| `Space` `l` `g` | Open the Git interface |

Requires the `lazygit` CLI: `brew install lazygit`.

### oil.nvim

| Key | Action |
|---|---|
| `-` | Open the directory editor |
| `i`, edit the filename, `Esc`, then `:w` | Rename a file |
| `dd`, then `:w` | Delete a file |

### General navigation

| Key | Action |
|---|---|
| `Ctrl-D` | Half-page scroll down |
| `Ctrl-U` | Half-page scroll up |
| `Ctrl-E` | Smooth scroll down |
| `Ctrl-Y` | Smooth scroll up |
| `:q` | Quit the current buffer |
| `:qa` | Quit all buffers |
| `:w` | Save the current buffer |
