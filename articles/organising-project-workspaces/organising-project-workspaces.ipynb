## 🎯 Problem Definition

When you're working on a project, you typically have the following things open:

1. A text editor.
2. A terminal emulator.
3. These days, an agent harness (e.g., Claude)

In popular editors like Visual Studio Code, all these three come built in. The terminal emulator is a built-in tool, and the agent harness is an official plugin developed by the harness providers themselves. 

How do we set our workflows using neovim in a way that it matches what we have in popular editors?

There are different ways of doing this, and one way might not fit you all. However, I can give you my own workflow, and you can customise it according to your style. I can give you some other suggestions as well so you can have a look and take some ideas in order to plug them into your own workflows.

## 🖥️ The Terminal Emulator

It is my opinion that when you are working on a project, you rarely have to leave the terminal. Unfortunately, this case breaks when you want to search the web, but other than that, the premise still holds true. The workflow that I have set up is purely terminal-based. So that is what I explain here.

Like I explained in [Day 01](https://dasunpubudumal.github.io/nvim-essentials/day-01/) of the series, you need to have a good terminal emulator. I highly recommend [Kitty](https://github.com/kovidgoyal/kitty) or [Ghostty](https://ghostty.org/). **I don't recommend any terminal that does not have configuration via dotfiles** (which means I don't recommend iTerm). Configuring a terminal should be external, and in this day and age where all tools are configured via dotfiles (and dotfiles are often version controlled), terminals that control their configuration purely via a UI is objectively out-of-date.

Another property that your terminal needs to have is good graphics support. Kitty, for example, has [its own graphics protocol](https://sw.kovidgoyal.net/kitty/graphics-protocol/), while many others extend from it. Mitchell Hashimoto's Ghostty also supports it. 

### 🪟 Terminal Multiplexers

Now, I've always been a huge fan of terminal multiplexers (particularly `tmux`). However, I got to research them a little bit and I understood later that they are actually trying to emulate a terminal inside a terminal emulator! This is resource heavy, and therefore I decided to use terminal-native features like [Kitty Sessions](https://sw.kovidgoyal.net/kitty/sessions/), and not use `tmux` anymore. 

The principle behind having `tmux` as part of your workflow is to have different `tmux` sessions and within each session, have different `tmux` panes. For example, if you're working on two projects, you would have two different `tmux` sessions, each with three panes: 1) the text editor, 2) terminal and 3) agent harness. The key idea is that they are organised, and switching between each of these sessions and panes are a bunch of hotkeys away. When you get used to it, it becomes second nature. If you want to try out the `tmux` configuration I used, visit my [dotfiles](https://github.com/dasunpubudumal/.tmux).

> 📝 **Note:** I do not use `tmux` anymore due to the reason explained above.

### 🧩 Terminal-native features

What I do recommend, however, is to use terminal-native features to have this organisation. Both Kitty and Ghostty have terminal tabs, but only Kitty has sessions (although not persistent; persistent sessions are part of a next release, as mentioned by [Kovid](https://github.com/kovidgoyal) himself in a podcast). So if you're on Ghostty, and you want to work on multiple projects, you would probably have to open multple terminals, and have the tabs we discussed within each terminal window.

### 🧭 tmux vs terminal-native features at a glance

The diagram below puts the two set-ups side by side. Both give you a session per project, each holding `nvim`, a shell and an agent harness. The difference is what sits between those programs and your terminal emulator. With `tmux` (A), a tmux server sits in the middle and emulates a terminal of its own, then redraws the result into the real emulator. With terminal-native features (B), every program talks to the emulator directly.

### 🐱 My Personal Setup

The way that I personally have set up Kitty is with heavy customisation. I have used several [kittens](https://sw.kovidgoyal.net/kitty/kittens_intro/#kittens) (i.e., custom Python scripts) to extend Kitty in a way that I myself want my tool be behave like. For example, I have a fuzzy search for projects configured using [`fzf`](https://github.com/junegunn/fzf), and when selected it opens a session for that project that contains three tabs: 1) Neovim opened in the project directory, 2) a terminal opened in the project directory, 3) Claude CLI opened in the project directory. I also have another kitten set up to select the sessions based on the sessions that I have currently open, and another for closing any session that I have open. I have mapped keys `cmd+shift+n`, `cmd+shift+s` and `cmd+shift+x` for these tasks respectively; in my Linux setup `cmd` is replaced with `ctrl` because `cmd` acts as `super` and Wayland gets conflicts with its own keybindings up when I use `super` inside the terminal.

#### 👻 The same set-up with Ghostty

Ghostty has tabs but no sessions, so there is nothing to group a project's tabs under. Instead, you open one Ghostty window per project and keep that project's `nvim`, shell and agent harness as tabs inside it. Switching tabs uses Ghostty's own tab keys, and switching projects means switching OS windows.

## 🤖 The Agent Harness

When working with Visual Studio Code, you probably interact with the installed agent harness using the agent plugin. You _can_ do that through neovim as well, but I do recommend using the agent's CLI tool. For Claude the CLI tool can be installed by following [this](https://code.claude.com/docs/en/quickstart#step-1-install-claude-code) guide. 

After installation, ideally, you could type `claude` after you navigate to your project directory inside the third terminal tab you have opened (explained in section above) for agents, and your agent should be able to understand the context of the project for you.

## 🏁 Conclusion

There is no single right way to organise project workspaces around Neovim, but the goal is always the same: get the editor, a terminal and an agent harness for each project within a few keystrokes of each other, the way VS Code puts them in one window. In this article, we covered:

1. **The problem.** Every project needs an editor, a terminal and (these days) an agent harness, and in a terminal-based workflow you have to organise them yourself.
2. **The terminal emulator.** Choose one that is configured through dotfiles and has good graphics support, such as Kitty or Ghostty.
3. **`tmux` vs terminal-native features.** `tmux` gives you sessions and panes, but it emulates a terminal inside your terminal. Kitty's sessions and tabs give you the same organisation without that extra layer.
4. **My personal setup.** Kitty kittens create a session per project with `IDE`, `term` and `Agents` tabs (`cmd+shift+n`), switch between open sessions (`cmd+shift+s`) and close them (`cmd+shift+x`).
5. **Ghostty.** Without sessions, use one window per project with the same three tabs inside it.
6. **The agent harness.** Run the agent's CLI (e.g., `claude`) from the project directory in its own tab.

Take whichever pieces suit you. Start simple, with one window or session per project and the same three tabs every time, and only automate it once the layout feels natural.
