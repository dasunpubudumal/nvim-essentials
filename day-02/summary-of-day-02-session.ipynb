## 🛠️ Technical Setup

### Opening Neovim

| Command | What it does |
|---|---|
| `nvim` | Start Neovim |
| `nvim <directory>` | Open a directory |
| `nvim .` | Open the current directory (run it from inside the project) |

### Installing the LazyVim starter

These steps follow the [LazyVim installation guide](https://www.lazyvim.org/installation).

**1. Back up any existing Neovim config** (skip this if `~/.config/nvim` doesn't exist):

```bash
mv ~/.config/nvim{,.bak}
```

**2. Clone the [starter](https://github.com/LazyVim/starter) into `~/.config/nvim`:**

```bash
git clone https://github.com/LazyVim/starter ~/.config/nvim
```

**3. Remove the starter's `.git` folder**, so that the config becomes your own rather than a copy of the starter repository:

```bash
rm -rf ~/.config/nvim/.git
```

**4. Start Neovim.** On the first launch, LazyVim downloads and installs its bundled plugins:

```bash
nvim
```

### Adding a plugin

1. Create a new `.lua` file in `~/.config/nvim/lua/plugins/` (for example, `lazygit.lua` or `oil.lua`).
2. Copy the plugin's configuration from its GitHub page. Make sure the file **returns a Lua table**, starting with `return`.
3. Restart Neovim. It picks up the new file and installs the plugin.

The two plugins from the session look like this:

```lua
-- ~/.config/nvim/lua/plugins/lazygit.lua
return {
  "kdheepak/lazygit.nvim",
  cmd = { "LazyGit" },
  dependencies = { "nvim-lua/plenary.nvim" },
  keys = {
    { "<leader>lg", "<cmd>LazyGit<cr>", desc = "LazyGit" },
  },
}
```

```lua
-- ~/.config/nvim/lua/plugins/oil.lua
return {
  "stevearc/oil.nvim",
  lazy = false,
  opts = {},
  dependencies = { "nvim-tree/nvim-web-devicons" },
  keys = {
    { "-", "<cmd>Oil<cr>", desc = "Open parent directory" },
  },
}
```

> 💡 `<leader>` is `Space` in LazyVim, so `<leader>lg` means `Space` then `l` then `g`.

## ⌨️ Key Bindings & Tools

### File explorer

| Key | Action |
|---|---|
| `Space` `e` | Open or close the file explorer panel |
| `h` / `j` / `k` / `l` | Move around in the explorer |

### LazyGit

| Key | Action |
|---|---|
| `Space` `l` `g` | Open the LazyGit interface |

LazyGit needs the `lazygit` CLI: install it with `brew install lazygit`.

### oil.nvim

oil shows a directory as an ordinary buffer: you edit the list of files like text, and saving applies the changes.

| Key | Action |
|---|---|
| `-` | Open the directory editor (the parent directory of the current file) |
| `i`, edit the name, `Esc`, then `:w` | Rename a file |
| `dd`, then `:w` | Delete a file |

### General navigation

| Key | Action |
|---|---|
| `Ctrl-D` | Scroll down half a page |
| `Ctrl-U` | Scroll up half a page |
| `Ctrl-E` | Scroll down (smoothly, one line at a time) |
| `Ctrl-Y` | Scroll up (smoothly, one line at a time) |
| `:w` | Save the current buffer |
| `:q` | Quit the current buffer |
| `:qa` | Quit all buffers |
