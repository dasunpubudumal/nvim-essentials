We went through how the Neovim workflow fits together (terminal tabs and sessions, macOS workspaces, and plugins such as oil.nvim and LazyVim), which terminal to use, how to use Claude alongside Neovim, and what Tree-sitter and LSPs each do. The full material for the day is in the [Day 4 notebook](day-04.ipynb), and the workspace setup is covered in more detail in the [Organising Project Workspaces](../articles/organising-project-workspaces/organising-project-workspaces.ipynb) article.

## ✅ Key Outcomes

- **The workflow is more than the editor.** The Neovim workflow combines terminal-level organization (tabs and sessions), macOS workspaces, and plugins such as oil.nvim and LazyVim, to keep hands off the mouse and work faster.
- **Terminal recommendation.** [Kitty](https://sw.kovidgoyal.net/kitty/) and [Ghostty](https://ghostty.org/) are recommended, because their configuration lives in files outside the app and they support graphics protocols that Neovim can use.
- **Claude through the CLI.** The official Claude CLI is the preferred agentic workflow, rather than a Neovim plugin, to avoid depending on a third party for the licence or tokens.
- **Work follow-up.** The ERP integration needs a check on the cost codes integration, specifically the missing budget holder field that Rich reported.

## 🧭 Decisions Made

| Topic | Decision |
|---|---|
| Terminal | Use Kitty or Ghostty rather than iTerm2. Their configuration is kept in dotfiles, so it can be version-controlled and rolled back easily. |
| Claude | Use the official Claude CLI rather than a Neovim plugin, to keep direct control of the licence and tokens. |
| Workspaces | Use macOS workspaces (`F3`) to keep applications apart, for example the browser on Desktop 1 and the terminal on Desktop 2, so you don't have to fight with window focus. |
| Project layout | Within a single project window, use terminal tabs (`Cmd` + number) to separate the editor, a shell, and the agent. |

## 🛠️ Technical Guidance

| Tool | What it is |
|---|---|
| [Tree-sitter](https://tree-sitter.github.io/tree-sitter/) | A project, used well beyond Neovim, that parses source code into an abstract syntax tree (AST). Neovim uses the tree for highlighting, go-to-definition and documentation. |
| LSP | A language server builds on the syntax tree to give a higher-level, semantic understanding of the code, such as type checking and suggestions. |
| [tmux](https://github.com/tmux/tmux) | Useful for sessions that persist, but it can slow things down. Dasun has moved away from it in favour of Kitty's own session management. |
| [oil.nvim](https://github.com/stevearc/oil.nvim) | A file explorer that lets you edit the file system as if it were a Vim buffer. |
