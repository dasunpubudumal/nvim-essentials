# Neovim Essentials: Day 1 - Look & Feel

> Neovim is a refactor, and sometimes redactor, in the tradition of Vim (which itself derives from [Stevie](https://en.wikipedia.org/wiki/Stevie_%28text_editor%29)) [...] Many clones and derivatives exist, some very clever—but none are Vim.
>
> Neovim is for users who want the good parts of Vim, and more.

**Note**: I use "nvim" as a shorthand to Neovim throughout this article.

## Goals of the Neovim Project

Extensible. Usable. Vim.

- Retain the character of Vim — fast, versatile quasi-minimal.
- Unblock plugin authors.
- Deliver a first-class [Lua](https://www.lua.org/) interface, as an alternative to Vimscript.
- Optimize “out of the box”, for new users but especially regular users.
- In matters of taste/ambiguity, favor tradition/compatibility…
- …but prefer usability if the benefits are extreme.

Source: [Neovim Charter](https://neovim.io/charter/)

## The mindset that Neovim requires

- Neovim is **not a tool "that just works"**. It's a tool that requires work from you.
- Raw nvim is **not** meant for you to use as it is. It is only a starting point.
- **You** need to make it yours. Nobody else can do it for you ([LazyVim](https://www.lazyvim.org/) probably comes close).
- To make it yours, you need to:
  - Spend time on it.
  - Research. I suggest keeping tabs on [r/neovim](https://www.reddit.com/r/neovim/) (Reddit), which is the best place for new plugins and community FAQ.
  - See how others have configured it, and **cherry-pick what you need/like**. Like said before, it might not suit you as a whole.
- Understand that it takes time to get to know it.
  - It requires a different way of looking at a piece of text.
  - You're not going to move inside nvim like an experienced person in a week. It takes ~3-4 weeks to just to get the motions into muscle memory.
  - So, be patient.
- The better you know your keyboard, better your nvim experience is going to be. So, **have a good keyboard and know it well**.
  - You need to reach different keys fast. Some keys in the keyboard are useless (e.g., Caps Lock). Map them to useful ones (e.g., Esc).
- You're **not going to use your mouse**.
  - In the beginning, it might tempting to use the mouse. But try to stay away from it.
  - If you're using your mouse in nvim, you've either missed/forgotten a config/keybinding, or configured something incorrectly.
- A good terminal makes all the difference. So, know your terminal.
  - Neovim plugins depend on certain features of terminals (terminal emulators). So, you need to have a good terminal.
  - Choose a terminal that has configuration externalised (i.e., configuration files), along with good docs on how to customise.
  - I recommend [Ghostty](https://ghostty.org/) (my favourite), [Kitty](https://github.com/kovidgoyal/kitty), or [iTerm2](https://iterm2.com/).
  - Some terminals mess with the `Meta` key that `tmux` uses. In macOS, it's `opt` key but I've noticed some terminal assigns this to other weird keys.
- Neovim alone wouldn't be enough. To have a ["flow"-like state](https://www.theguardian.com/science/article/2024/jul/20/flow-state-science-creativity-psychology-focus), you would most definitely need to couple nvim with a terminal multiplexer.
  - There are a couple of good ones ([`tmux`](https://github.com/tmux/tmux/wiki) leads the battle with [`zellij`](https://zellij.dev/) not far behind)
  - I use `tmux`. So, we'll cover how we couple `tmux` with nvim in a future session (most probably Day 3).

## Installing Neovim

### Step 1: Check if it's already installed

Before installing anything, check whether nvim is already on your machine:

```bash
nvim --version
```

- If you see something like `NVIM v0.x.y` followed by build details, it's already installed. Note the version number.
- If you get `command not found: nvim`, it's not installed (or not on your `PATH`). Move on to Step 2.

You can also see *where* it's installed from (useful if you have more than one copy):

```bash
which -a nvim
```

**Is your version recent enough?** Compare it against the latest stable release on the [Neovim releases page](https://github.com/neovim/neovim/releases). Many plugins (and distributions like LazyVim) require a recent version, and system package managers — especially on Linux — often ship old ones. If yours is several minor versions behind, upgrade using one of the methods below.

### Step 2: Install (or upgrade)

**macOS** (with [Homebrew](https://brew.sh/)):

```bash
brew install neovim      # fresh install
brew upgrade neovim      # if already installed via Homebrew
```

**Linux**:

Your distro's package manager may give you an outdated version, so prefer the official pre-built release:

```bash
curl -LO https://github.com/neovim/neovim/releases/latest/download/nvim-linux-x86_64.tar.gz
sudo rm -rf /opt/nvim-linux-x86_64
sudo tar -C /opt -xzf nvim-linux-x86_64.tar.gz
```

Then add it to your `PATH` (e.g. in `~/.bashrc` or `~/.zshrc`):

```bash
export PATH="$PATH:/opt/nvim-linux-x86_64/bin"
```

(On ARM machines, replace `x86_64` with `arm64`.)

**Windows**:

```powershell
winget install Neovim.Neovim
```

For other options (AppImage, Snap, Scoop, building from source, etc.), see the official [installation guide](https://github.com/neovim/neovim/blob/master/INSTALL.md).

### Step 3: Verify

Open a **new** terminal and run the check again:

```bash
nvim --version
```

You should now see the version you just installed. Launch it with `nvim`, and quit with `:q` followed by `Enter`.

## LazyVim

### What is it?

[LazyVim](https://www.lazyvim.org/) is a **Neovim distribution**: a ready-made configuration that turns plain nvim into a full IDE-like editor out of the box. It was created by [folke](https://github.com/folke), who also wrote many popular Neovim plugins.

It gives you, preconfigured:

- A file explorer, fuzzy finder (files, text, buffers), and a start dashboard.
- Syntax highlighting via Treesitter, and LSP support (autocompletion, go-to-definition, diagnostics) with language servers installed through [Mason](https://github.com/mason-org/mason.nvim).
- Git integration, a status line, keymap hints (press `<Space>` and wait), and a polished theme.
- **Language extras**: one-line opt-ins for Python, Rust, Go, etc., via the `:LazyExtras` menu.

A few things that often cause confusion:

- **LazyVim ≠ lazy.nvim.** [lazy.nvim](https://github.com/folke/lazy.nvim) is the *plugin manager* that installs and loads plugins. LazyVim is a *configuration* built on top of it.
- **It's still your config.** LazyVim lives in `~/.config/nvim` like any other config. You can override any default and add your own plugins in `lua/plugins/`.
- **It's a good starting point, not the finish line.** As mentioned above, nvim is something you make your own. LazyVim gets you a usable editor on day one while you learn what you actually want.

### Requirements

- A recent Neovim. Check the exact minimum version in the [LazyVim installation docs](https://www.lazyvim.org/installation), and compare it with `nvim --version`.
- `git`
- A [Nerd Font](https://www.nerdfonts.com/) set as your terminal font (optional, but needed for icons).
- A C compiler (for Treesitter). On macOS, `xcode-select --install` provides one.
- Recommended: `ripgrep`, `fd`, `fzf` (for searching) and `lazygit` (git UI). On macOS:

```bash
brew install git ripgrep fd fzf lazygit
```

### Step 1: Check for an existing config

LazyVim installs into `~/.config/nvim`. If you already have something there, back it up first rather than overwriting it:

```bash
ls ~/.config/nvim
```

- `No such file or directory`: you have no config yet. Skip to Step 2.
- Files are listed: back them up, along with nvim's data, state and cache folders (these hold plugins and history tied to the old config):

```bash
mv ~/.config/nvim{,.bak}

# optional but recommended
mv ~/.local/share/nvim{,.bak}
mv ~/.local/state/nvim{,.bak}
mv ~/.cache/nvim{,.bak}
```

### Step 2: Install the LazyVim starter

```bash
git clone https://github.com/LazyVim/starter ~/.config/nvim
rm -rf ~/.config/nvim/.git
```

Removing `.git` disconnects the folder from the starter repository, so you can put it under your own version control later.

### Step 3: Launch and check health

```bash
nvim
```

On first launch, lazy.nvim downloads and installs all the plugins. Wait for it to finish, then restart nvim. Afterwards, run:

```vim
:LazyHealth
```

This checks that everything (Neovim version, git, compiler, ripgrep, etc.) is in place, and tells you what's missing.

### Useful commands to start with

| Command / key | What it does |
|---|---|
| `<Space>` (wait) | Shows available keymaps |
| `:Lazy` | Plugin manager UI (update, inspect, clean plugins) |
| `:LazyExtras` | Enable language support and optional features |
| `:Mason` | Install language servers, formatters and linters |
| `<Space><Space>` | Find files |
| `<Space>/` | Search text in the project |
| `<Space>e` | Toggle file explorer |

## In the meantime..

- Try to make Neovim your primary editor. It will be difficult, but when you need to open an editor, open Neovim and try to do your work there instead of your comfort-editor.
  - **It’s okay to struggle**. Growth begins where comfort ends; all you have to do is make it through. If learning feels effortless, you’re probably not really growing.
